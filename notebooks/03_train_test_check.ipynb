```python
import joblib, pandas as pd

logreg = joblib.load("data/processed/models/logistic.joblib")
scaler = joblib.load("data/processed/models/scaler.joblib")
xgb = joblib.load("data/processed/models/xgboost.joblib")

X_test = pd.read_csv("data/processed/models/X_test.csv")
y_test = pd.read_csv("data/processed/models/y_test.csv").squeeze()

print("Logistic:", logreg.score(scaler.transform(X_test), y_test))
print("XGBoost :", xgb.score(X_test, y_test))
```

```python
for feat, coef in zip(X_test.columns, logreg.coef_[0]):
    print(f"{feat:12s} {coef:+.3f}")
```
