```python
# how many test games exceed 38 frames?
import numpy as np
df = pd.read_csv("data/processed/features.csv")
test_ids = np.load("data/processed/models/test_ids.npy", allow_pickle=True)
lengths = df[df["match_id"].isin(test_ids)].groupby("match_id").size()
print("games > 38 frames:", (lengths > 38).sum())
print("frames lost to truncation:", (lengths[lengths > 38] - 38).sum())
```
