```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()
API_KEY = os.getenv("RIOT_API_KEY")
headers = {"X-Riot-Token": API_KEY}

url = "https://kr.api.riotgames.com/lol/league/v4/challengerleagues/by-queue/RANKED_SOLO_5x5"
response = requests.get(url, headers=headers)

print(response.status_code)          # confirm 200 first
data = response.json()
print(data.keys())                   # confirm top-level shape
print(data["entries"][0])            # inspect one entry's actual fields
```
